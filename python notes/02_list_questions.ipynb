# Python Practice – Filtering & Transforming Lists
Each question is solved in two ways:
- **Method 1:** `for` loop + `if` + `append()`
- **Method 2:** List comprehension

### Q1. Find numbers greater than 50 and store all the numbers into new list.

```python
numbers = [20, 55, 30, 80, 45, 90, 10]

# Method 1
result = []
for n in numbers:
    if n > 50:
        result.append(n)
print("Method 1:", result)

# Method 2
result = [n for n in numbers if n > 50]
print("Method 2:", result)
```

### Q2. Find numbers between 20 and 50 and store all the numbers into new list.

```python
numbers = [10, 25, 35, 60, 45, 15, 50, 70]

# Method 1
result = []
for n in numbers:
    if 20 <= n <= 50:
        result.append(n)
print("Method 1:", result)

# Method 2
result = [n for n in numbers if 20 <= n <= 50]
print("Method 2:", result)
```

### Q3. Find numbers divisible by both 3 and 5 and store all the numbers into new list.

```python
numbers = [10, 15, 20, 30, 45, 50, 60, 75]

# Method 1
result = []
for n in numbers:
    if n % 3 == 0 and n % 5 == 0:
        result.append(n)
print("Method 1:", result)

# Method 2
result = [n for n in numbers if n % 3 == 0 and n % 5 == 0]
print("Method 2:", result)
```

### Q4. Find names having more than 4 characters and store result into new list.

```python
names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]

# Method 1
result = []
for name in names:
    if len(name) > 4:
        result.append(name)
print("Method 1:", result)

# Method 2
result = [name for name in names if len(name) > 4]
print("Method 2:", result)
```

### Q5. Find names starting with "A" and store result into new list.

```python
names = ["Amit", "Rahul", "Ankit", "Neha", "Arjun", "Rohit"]

# Method 1
result = []
for name in names:
    if name.startswith("A"):
        result.append(name)
print("Method 1:", result)

# Method 2
result = [name for name in names if name.startswith("A")]
print("Method 2:", result)
```

### Q6. Extract vowels from a string and store result into new list.

```python
text = "programming"

# Method 1
result = []
for ch in text:
    if ch in "aeiouAEIOU":
        result.append(ch)
print("Method 1:", result)

# Method 2
result = [ch for ch in text if ch in "aeiouAEIOU"]
print("Method 2:", result)
```

### Q7. Extract digits from a string and store result into new list.

```python
data = "Python123Programming456"

# Method 1
result = []
for ch in data:
    if ch.isdigit():
        result.append(ch)
print("Method 1:", result)

# Method 2
result = [ch for ch in data if ch.isdigit()]
print("Method 2:", result)
```

### Q8. Replace every even number with "Even" and every odd number with "Odd" and store result into new list.
Expected Output: `['Even', 'Odd', 'Even', 'Odd', 'Even']`

```python
numbers = [10, 15, 20, 25, 30]

# Method 1
result = []
for n in numbers:
    if n % 2 == 0:
        result.append("Even")
    else:
        result.append("Odd")
print("Method 1:", result)

# Method 2
result = ["Even" if n % 2 == 0 else "Odd" for n in numbers]
print("Method 2:", result)
```

### Q9. Replace negative numbers with 0 and store result into new list.
Expected Output: `[10, 0, 20, 0, 30, 0]`

```python
numbers = [10, -5, 20, -8, 30, -2]

# Method 1
result = []
for n in numbers:
    if n < 0:
        result.append(0)
    else:
        result.append(n)
print("Method 1:", result)

# Method 2
result = [0 if n < 0 else n for n in numbers]
print("Method 2:", result)
```

### Q10. Give a 10% discount to products costing ₹1000 or more, otherwise give 5% and store result into new list.
Expected Output: `[475.0, 1080.0, 760.0, 1800.0, 1350.0]`

```python
prices = [500, 1200, 800, 2000, 1500]

# Method 1
result = []
for p in prices:
    if p >= 1000:
        result.append(p * 0.90)
    else:
        result.append(p * 0.95)
print("Method 1:", result)

# Method 2
result = [p * 0.90 if p >= 1000 else p * 0.95 for p in prices]
print("Method 2:", result)
```

### Q15. Bank Transactions: Positive values represent deposits. Find all deposits and all withdrawals.
Expected Output (deposits): `[5000, 3000, 7000]`
Expected Output (withdrawals): `[-2000, -1000, -500]`

```python
transactions = [5000, -2000, 3000, -1000, 7000, -500]

# Method 1
deposits = []
withdrawals = []
for t in transactions:
    if t > 0:
        deposits.append(t)
    else:
        withdrawals.append(t)
print("Method 1 - Deposits:   ", deposits)
print("Method 1 - Withdrawals:", withdrawals)

# Method 2
deposits = [t for t in transactions if t > 0]
withdrawals = [t for t in transactions if t < 0]
print("Method 2 - Deposits:   ", deposits)
print("Method 2 - Withdrawals:", withdrawals)
```

### Q16. Find names of students who scored 60 or more.

### Q18. Find employees who earn more than ₹60,000 and belong to IT.

```python
employees = [
    {"name": "Amit", "department": "IT", "salary": 55000},
    {"name": "Rahul", "department": "HR", "salary": 70000},
    {"name": "Neha", "department": "IT", "salary": 80000},
    {"name": "Rohit", "department": "IT", "salary": 45000}
]

# Method 1
result = []
for emp in employees:
    if emp["salary"] > 60000 and emp["department"] == "IT":
        result.append(emp["name"])
print("Method 1:", result)

# Method 2
result = [emp["name"] for emp in employees
          if emp["salary"] > 60000 and emp["department"] == "IT"]
print("Method 2:", result)
```
